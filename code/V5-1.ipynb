## ⚙️ CELDA 1: Configuración Global y Setup

```python
import os
import sys
import json
import time
import pickle
import dill
import random
import itertools
import numpy as np
import pandas as pd
import torch
import selfies as sf
from rdkit import Chem
from tqdm import tqdm
from math import ceil
from collections import OrderedDict

# ─────────────────────────────────────────
# DEVICE DETECTION
# ─────────────────────────────────────────
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

if torch.cuda.is_available():
    vram_gb = torch.cuda.get_device_properties(0).total_memory / 1e9
    gpu_name = torch.cuda.get_device_name(0)
    print(f"🖥️  GPU detectada: {gpu_name}")
    print(f"💾 VRAM disponible: {vram_gb:.1f} GB")
    if vram_gb >= 10:
        recommended_esm  = "esm2_t33_650M_UR50D"
        recommended_chem = "DeepChem/ChemBERTa-77M-MTR"
        print("🏆 Configuración RECOMENDADA: ESM-2 650M + ChemBERTa-77M")
    elif vram_gb >= 6:
        recommended_esm  = "esm2_t30_150M_UR50D"
        recommended_chem = "DeepChem/ChemBERTa-77M-MTR"
        print("✅ Configuración MEDIA: ESM-2 150M + ChemBERTa-77M")
    else:
        recommended_esm  = "esm2_t6_8M_UR50D"
        recommended_chem = "DeepChem/ChemBERTa-10M-MTR"
        print("🔋 Configuración LIGERA: ESM-2 8M + ChemBERTa-10M (CPU compatible)")
else:
    print("⚠️  Sin GPU — usando CPU. Extracción de features será lenta.")
    recommended_esm  = "esm2_t6_8M_UR50D"
    recommended_chem = "DeepChem/ChemBERTa-10M-MTR"
    vram_gb = 0

print(f"\n📍 Dispositivo activo: {DEVICE}")

# ─────────────────────────────────────────
# RUTAS
# ─────────────────────────────────────────
%cd /home/eag/Documents/DTIAM-YOSH/DTIAM-ModStack/code/

master_save_path = "../data/master_features/"
os.makedirs(master_save_path, exist_ok=True)
os.makedirs("../results/", exist_ok=True)
os.makedirs("../models/trained/", exist_ok=True)

# ══════════════════════════════════════════════════════════════════
# ✅ OPCIÓN A: NORMALIZACIÓN DE SMILES CON RDKIT
# ══════════════════════════════════════════════════════════════════
def canonicalize_smiles(smi: str) -> str | None:
    """
    Canonicaliza un SMILES usando RDKit.
    
    Beneficios:
    - Garantiza unicidad: 'CC(O)' y 'OC(C)' → mismo SMILES canónico
    - Descarta SMILES inválidos (retorna None) en lugar de fallar silenciosamente
    - Normaliza cargas, aromaticidad y estereoquímica de forma consistente
    
    Returns:
        SMILES canónico (str) o None si el input es inválido
    """
    if not smi or not isinstance(smi, str):
        return None
    try:
        mol = Chem.MolFromSmiles(smi.strip())
        return Chem.MolToSmiles(mol) if mol is not None else None
    except Exception:
        return None


# ══════════════════════════════════════════════════════════════════
# ✅ OPCIÓN B: CONVERSIÓN SMILES → SELFIES
# ══════════════════════════════════════════════════════════════════
def smiles_to_selfies(smi: str) -> str | None:
    """
    Convierte SMILES a SELFIES.
    
    SELFIES garantiza que CUALQUIER string SELFIES generado represente
    una molécula químicamente válida, eliminando el problema de SMILES
    inválidos o mal formados.
    
    Internamente canonicaliza el SMILES primero (Opción A) para que la
    conversión sea reproducible.
    
    Returns:
        String SELFIES (str) o None si el SMILES no puede parsearse
    """
    canon = canonicalize_smiles(smi)
    if canon is None:
        return None
    try:
        return sf.encoder(canon)
    except sf.exceptions.EncoderError:
        return None


def selfies_to_smiles(sel: str) -> str | None:
    """Convierte SELFIES → SMILES canónico (útil para visualización con RDKit)."""
    try:
        raw = sf.decoder(sel)
        return canonicalize_smiles(raw)
    except Exception:
        return None


def normalize_dataframe(
    df: pd.DataFrame,
    smi_col: str = "smi",
    mode: str = "canonical_smiles"   # "canonical_smiles" | "selfies"
) -> pd.DataFrame:
    """
    Normaliza la columna de SMILES de un DataFrame.
    
    Opción A (mode='canonical_smiles'):
        - Canonicaliza cada SMILES con RDKit
        - Elimina filas con SMILES inválidos
        - Elimina duplicados reales (mismo compuesto, SMILES distintos)
    
    Opción B (mode='selfies'):
        - Canonicaliza SMILES primero
        - Convierte a SELFIES
        - Agrega columna 'selfies' al DataFrame
        - Elimina filas donde la conversión falló
    
    Returns:
        DataFrame limpio con columna smi (y 'selfies' en modo B)
    """
    original_len = len(df)
    df = df.copy()
    
    if mode == "canonical_smiles":
        print(f"   🔧 Opción A: Canonicalizando {original_len} SMILES con RDKit...")
        df[smi_col] = df[smi_col].apply(canonicalize_smiles)
        df = df.dropna(subset=[smi_col])
        # Eliminar duplicados REALES (mismo SMILES canónico = mismo compuesto)
        df = df.drop_duplicates(subset=[smi_col])
        invalids = original_len - len(df)
        print(f"   ✅ {len(df)} compuestos únicos válidos "
              f"({invalids} inválidos/duplicados reales eliminados)")
    
    elif mode == "selfies":
        print(f"   🔧 Opción B: Convirtiendo {original_len} SMILES → SELFIES...")
        df[smi_col] = df[smi_col].apply(canonicalize_smiles)  # A primero
        df = df.dropna(subset=[smi_col])
        df["selfies"] = df[smi_col].apply(smiles_to_selfies)
        df = df.dropna(subset=["selfies"])
        # Duplicados: mismo SELFIES = mismo compuesto
        df = df.drop_duplicates(subset=["selfies"])
        invalids = original_len - len(df)
        print(f"   ✅ {len(df)} compuestos únicos en SELFIES "
              f"({invalids} fallidos/duplicados eliminados)")
    
    return df.reset_index(drop=True)


# ─────────────────────────────────────────
# CATÁLOGO DE PLMs DISPONIBLES
# ─────────────────────────────────────────
PLM_CATALOG = {
    # ── Compound PLMs ─────────────────────────────────────────────────
    "compound": {
        # ── Opción A: ChemBERTa (input: SMILES canónico) ──────────────
        "chemberta_light": {
            "name": "DeepChem/ChemBERTa-10M-MTR",
            "params": "10M",
            "vram_req": "<2GB",
            "embed_dim": 384,
            "input_format": "smiles",      # ← Opción A
            "description": "ChemBERTa ligero — CPU/VRAM mínima"
        },
        "chemberta_medium": {
            "name": "DeepChem/ChemBERTa-77M-MTR",
            "params": "77M",
            "vram_req": "~3GB",
            "embed_dim": 768,
            "input_format": "smiles",      # ← Opción A
            "description": "ChemBERTa medium — balance óptimo (RECOMENDADO)"
        },
        "chemberta_base": {
            "name": "seyonec/ChemBERTa-zinc-base-v1",
            "params": "86M",
            "vram_req": "~3GB",
            "embed_dim": 768,
            "input_format": "smiles",      # ← Opción A
            "description": "ChemBERTa base — preentrenado en ZINC"
        },
        # ── Opción B: SELFormer (input: SELFIES) ───────────────────────
        "selformer": {
            "name": "HUBioinfo/SELFormer",
            "params": "86M",
            "vram_req": "~3GB",
            "embed_dim": 768,
            "input_format": "selfies",     # ← Opción B
            "description": "SELFormer — RoBERTa preentrenado en SELFIES. "
                           "Garantía 100% de moléculas válidas."
        },
    },
    # ── Protein PLMs (ESM-2) ───────────────────────────────────────────
    "protein": {
        "light": {
            "name": "esm2_t6_8M_UR50D",
            "params": "8M",
            "vram_req": "<2GB",
            "embed_dim": 320,
            "description": "CPU compatible, ideal para pruebas"
        },
        "medium": {
            "name": "esm2_t30_150M_UR50D",
            "params": "150M",
            "vram_req": "~4GB",
            "embed_dim": 640,
            "description": "Buen compromiso calidad/costo computacional"
        },
        "large": {
            "name": "esm2_t33_650M_UR50D",
            "params": "650M",
            "vram_req": "~8GB",
            "embed_dim": 1280,
            "description": "Máxima calidad, requiere GPU potente"
        }
    }
}

# Aliases cortos para compatibilidad con código anterior
PLM_CATALOG["compound"]["light"]  = PLM_CATALOG["compound"]["chemberta_light"]
PLM_CATALOG["compound"]["medium"] = PLM_CATALOG["compound"]["chemberta_medium"]
PLM_CATALOG["compound"]["base"]   = PLM_CATALOG["compound"]["chemberta_base"]

print("\n📋 PLM_CATALOG cargado:")
print("   Compuestos (Opción A — SMILES canónico):")
for k in ["chemberta_light", "chemberta_medium", "chemberta_base"]:
    c = PLM_CATALOG["compound"][k]
    print(f"   • {k:22s} {c['params']:5s} | {c['vram_req']:6s} | dim={c['embed_dim']}")
print("   Compuestos (Opción B — SELFIES):")
c = PLM_CATALOG["compound"]["selformer"]
print(f"   • selformer            {c['params']:5s} | {c['vram_req']:6s} | dim={c['embed_dim']}")
print("   Proteínas (ESM-2):")
for k, c in PLM_CATALOG["protein"].items():
    print(f"   • {k:22s} {c['params']:5s} | {c['vram_req']:6s} | dim={c['embed_dim']}")
```

## 🧪 CELDA 2: Extracción de Features de Compuestos

### Opción A — ChemBERTa con SMILES canónico (default)
Canonicaliza cada SMILES con RDKit antes de tokenizar. Resuelve duplicados y SMILES inválidos.

### Opción B — SELFormer con SELFIES
Convierte SMILES→SELFIES automáticamente. Usa `HUBioinfo/SELFormer`, un RoBERTa preentrenado sobre SELFIES.
Garantía 100% de que cada token representa una molécula válida.

La función `cal_comp_feat()` detecta automáticamente qué opción usar según `model_key`:
- `model_key` = `'chemberta_light'` / `'chemberta_medium'` / `'chemberta_base'` → **Opción A**
- `model_key` = `'selformer'` → **Opción B** (convierte SMILES→SELFIES internamente)


```python
from transformers import AutoTokenizer, AutoModel

def cal_comp_feat(
    data: pd.DataFrame,
    model_key: str = "chemberta_medium",
    batch_size: int = 64,
    pooling: str = "cls"
) -> dict:
    """
    Extrae embeddings de compuestos. Unifica Opción A (ChemBERTa/SMILES)
    y Opción B (SELFormer/SELFIES) en una sola función.

    Args:
        data:       DataFrame con columnas ['cid', 'smi']
        model_key:  Clave del PLM_CATALOG. Opciones:
                    Opción A → 'chemberta_light' | 'chemberta_medium' | 'chemberta_base'
                    Opción B → 'selformer'
        batch_size: Compuestos por batch (bajar si hay OOM).
        pooling:    'cls' (token [CLS]) | 'mean' (promedio de tokens)

    Returns:
        dict {cid: np.array(embed_dim)}
    """
    cfg = PLM_CATALOG["compound"][model_key]
    model_name   = cfg["name"]
    input_format = cfg["input_format"]  # 'smiles' | 'selfies'

    opcion = "B — SELFormer/SELFIES" if input_format == "selfies" else "A — ChemBERTa/SMILES canónico"
    print(f"\n🧪 Modelo: {model_name}  [{opcion}]")
    print(f"   Parámetros: {cfg['params']} | VRAM: {cfg['vram_req']} | dim={cfg['embed_dim']} | pooling={pooling}")

    # ── 1. Preparar y normalizar datos ──────────────────────────────────
    comp_data = data[["cid", "smi"]].copy()

    if input_format == "selfies":
        comp_data = normalize_dataframe(comp_data, mode="selfies")
        input_col = "selfies"   # La columna de texto que se tokenizará
    else:
        comp_data = normalize_dataframe(comp_data, mode="canonical_smiles")
        input_col = "smi"

    comp_data = comp_data.reset_index(drop=True)
    total = len(comp_data)

    if total == 0:
        print("❌ No quedaron compuestos válidos tras la normalización.")
        return {}

    # ── 2. Cargar modelo ────────────────────────────────────────────────
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model     = AutoModel.from_pretrained(model_name)
    if DEVICE.type == "cuda":
        model = model.half()
    model = model.to(DEVICE).eval()

    comp_feat  = {}
    error_cids = []
    start_time = time.time()
    TOTAL_ICONS = 20

    # ── 3. Inferencia por batches ────────────────────────────────────────
    for batch_start in range(0, total, batch_size):
        batch_df  = comp_data.iloc[batch_start : batch_start + batch_size]
        cids      = batch_df["cid"].tolist()
        texts     = batch_df[input_col].tolist()   # SMILES canónico o SELFIES

        try:
            encoded = tokenizer(
                texts,
                padding=True,
                truncation=True,
                max_length=512,
                return_tensors="pt"
            )
            input_ids      = encoded["input_ids"].to(DEVICE)
            attention_mask = encoded["attention_mask"].to(DEVICE)

            with torch.no_grad():
                outputs = model(input_ids=input_ids, attention_mask=attention_mask)

            hidden = outputs.last_hidden_state  # (B, seq_len, dim)

            if pooling == "cls":
                embeddings = hidden[:, 0, :]
            else:  # mean pooling con mask
                mask_exp   = attention_mask.unsqueeze(-1).float()
                embeddings = (hidden * mask_exp).sum(1) / mask_exp.sum(1).clamp(min=1e-9)

            embeddings_np = embeddings.cpu().float().numpy()
            for cid, emb in zip(cids, embeddings_np):
                comp_feat[cid] = emb

        except Exception as e:
            error_cids.extend(cids)

        finally:
            for var in ["input_ids", "attention_mask", "outputs", "hidden", "embeddings"]:
                if var in locals(): del locals()[var]
            torch.cuda.empty_cache()

        # Progreso
        current  = min(batch_start + batch_size, total)
        ratio    = current / total
        filled   = ceil(ratio * TOTAL_ICONS)
        icon_ok  = "🧬" if input_format == "selfies" else "💊"
        bar      = icon_ok * filled + "🧪" * (TOTAL_ICONS - filled)
        elapsed  = time.time() - start_time
        speed    = current / elapsed if elapsed > 0 else 0
        print(f"\r{int(ratio*100)}% |{bar}| {current}/{total} [{elapsed:.0f}s, {speed:.1f}cmp/s]", end="")

    model.cpu()
    del model, tokenizer
    torch.cuda.empty_cache()

    print(f"\n✅ ¡Compuestos listos! {len(comp_feat)}/{total} procesados en {time.time()-start_time:.1f}s")
    if error_cids:
        print(f"⚠️  {len(error_cids)} compuestos con error de tokenización")

    return comp_feat


# Alias para compatibilidad con V4
cal_comp_feat_chemberta = lambda data, model_key="chemberta_medium", **kw: cal_comp_feat(data, model_key, **kw)


# ══════════════════════════════════════════════════════════════════
# CONSOLIDACIÓN DE DATOS
# ══════════════════════════════════════════════════════════════════
print("\n--- CONSOLIDANDO DATOS ---")

datasets_info = [
    ("dti", "yamanishi_08", "drug_smiles.csv", "protein_seq.csv", ["cid", "smi"], ["pid", "seq"]),
    ("dti", "hetionet",     "drug_smiles.csv", "protein_seq.csv", ["cid", "smi"], ["pid", "seq"]),
    ("moa", "activation",  "drug_smi.csv",    "tar_seq.csv",     ["cid", "smi"], ["pid", "seq"]),
    ("moa", "inhibition",  "drug_smi.csv",    "tar_seq.csv",     ["cid", "smi"], ["pid", "seq"]),
]

all_drugs_data, all_targets_data = [], []

for task, ds, d_file, t_file, d_cols, t_cols in datasets_info:
    base_path = f"../data/{task}/{ds}/"
    if os.path.exists(base_path + d_file):
        df_d = pd.read_csv(base_path + d_file, sep="\t")
        df_d.columns = d_cols; all_drugs_data.append(df_d)
    if os.path.exists(base_path + t_file):
        df_t = pd.read_csv(base_path + t_file, sep="\t")
        df_t.columns = t_cols; all_targets_data.append(df_t)

for ds in ["davis", "kiba"]:
    base_path = f"../data/dta/{ds}/"
    if os.path.exists(base_path + "ligands_can.txt"):
        with open(base_path + "ligands_can.txt") as f:
            ligands = json.load(f, object_pairs_hook=OrderedDict)
            all_drugs_data.append(pd.DataFrame(ligands.items(), columns=["cid", "smi"]))
    if os.path.exists(base_path + "proteins.txt"):
        with open(base_path + "proteins.txt") as f:
            proteins = json.load(f, object_pairs_hook=OrderedDict)
            all_targets_data.append(pd.DataFrame(proteins.items(), columns=["pid", "seq"]))

# Concatenar crudo (sin dedup todavía — la dedup real la hace normalize_dataframe)
master_drugs_df_raw = pd.concat(all_drugs_data).reset_index(drop=True)
master_targets_df   = pd.concat(all_targets_data).drop_duplicates(subset=["pid"]).reset_index(drop=True)

print(f"🔬 Compuestos (antes de normalizar): {len(master_drugs_df_raw)}")
print(f"🧬 Proteínas únicas: {len(master_targets_df)}")
print()
print("💡 Elige el modo de representación molecular:")
print("   COMPOUND_MODEL_KEY = 'chemberta_medium'  → Opción A (SMILES canónico)")
print("   COMPOUND_MODEL_KEY = 'selformer'         → Opción B (SELFIES)")

# ══════════════════════════════════════════════════════════════════
# EJECUTAR — cambia COMPOUND_MODEL_KEY según tu elección
# ══════════════════════════════════════════════════════════════════
COMPOUND_MODEL_KEY = "chemberta_medium"   # ← Cambiar aquí: 'selformer' para Opción B

master_comp_feat = cal_comp_feat(
    master_drugs_df_raw,
    model_key=COMPOUND_MODEL_KEY,
    batch_size=64,
    pooling="cls"
)

# Guardar con nombre que incluye el modelo usado
feat_fname = f"MASTER_compound_features_{COMPOUND_MODEL_KEY}.pkl"
with open(master_save_path + feat_fname, "wb") as f:
    dill.dump(master_comp_feat, f)
print(f"💾 Guardado: {feat_fname}")
```
